# Manual PyTorch PINNs versus DeepXDE

This notebook compares the primary from-scratch PyTorch implementation with an optional framework-based implementation using DeepXDE. The purpose is not to claim that one approach is universally better. It is to identify what a framework abstracts and why manual implementation remains valuable for learning and customisation.

DeepXDE supports physics-informed learning workflows and multiple tensor backends, including PyTorch [1]. Its official Burgers demo uses geometry/time objects, boundary and initial-condition objects, and automatic derivative helpers [2].

[1]: https://deepxde.readthedocs.io/en/latest/ "DeepXDE documentation"
[2]: https://deepxde.readthedocs.io/en/latest/demos/pinn_forward/burgers.html "DeepXDE Burgers equation demo"

## What the manual implementation makes explicit

| Manual PyTorch file | Responsibility you can inspect directly |
|---|---|
| `src/pinn/core.py` | MLP, `torch.autograd`, loss loop, optimiser step |
| `src/problems/odes.py` | Cooling and oscillator residuals/conditions |
| `src/problems/burgers.py` | PDE residual, conditions, finite-difference reference |
| `experiments/train_*.py` | Collocation grid, run configuration, evaluation |

The manual implementation is the primary code in this repository. It helps you see every PINN mechanism.


## What DeepXDE abstracts

DeepXDE lets you declare a geometry, a PDE callback, condition objects, a `TimePDE`/`PDE` data object, a network, and a model. It handles much of the point sampling and training workflow. This reduces setup code, but it also means less of the mechanism is visible in your own source file.

## Comparison guide

| Criterion | Manual PyTorch | DeepXDE |
|---|---|---|
| Code volume | More code because each mechanism is explicit | Less user code for standard formulations |
| Learning value | Excellent for understanding residuals, derivatives, and loss construction | Excellent for understanding framework conventions and rapid setup |
| Flexibility | Direct control of every tensor operation and custom loss | High-level abstractions plus framework-supported features |
| Sampling and conditions | You build tensors and condition functions | Geometry and condition objects manage much of the setup |
| Evaluation plots | You create your own plotting functions | Framework provides helpers, but custom plots are still useful |
