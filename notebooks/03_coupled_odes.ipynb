# Coupled ODEs: damped harmonic oscillator

This notebook extends the PINN from a scalar output to two connected outputs. The network receives time and predicts position and velocity together:

$$t\longmapsto[x(t),v(t)].$$

The governing equations are:

$$\frac{dx}{dt}=v,$$

$$\frac{dv}{dt}=-2\zeta v-\omega^2x.$$

## Two outputs, two residuals

The model has `output_dim=2`. In `oscillator_residual`, PyTorch differentiates output column zero to obtain `dx/dt` and output column one to obtain `dv/dt`. The two residuals are then combined into one tensor for the generic trainer.


````python
import numpy as np
from problems.odes import OscillatorConfig, oscillator_reference

config = OscillatorConfig()
time = np.linspace(0.0, config.final_time, 10)
reference = oscillator_reference(time, config)
print("Reference state shape [time, position/velocity]:", reference.shape)
print("Initial state [x, v]:", reference[0])
````
