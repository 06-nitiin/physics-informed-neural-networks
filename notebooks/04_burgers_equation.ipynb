This notebook covers the nonlinear PDE challenge in the project:

$$u_t+u u_x=\nu u_{xx}. $$

The input now contains both position and time, and the network output is the field value `u(x, t)`.

## Inputs, derivatives, and conditions

The network maps `(x, t) → u`. The residual requires a time derivative, a first spatial derivative, and a second spatial derivative. The model is constrained by:

$$u(x,0)=-\sin(\pi x),$$

with zero values at both spatial boundaries.


````python
import numpy as np
from problems.burgers import BurgersConfig, burgers_initial_profile, burgers_reference

config = BurgersConfig()
x = np.linspace(config.x_min, config.x_max, 9)
print("Initial profile:", burgers_initial_profile(x))

x_ref, t_ref, reference = burgers_reference(config, nx=31, nt=31)
print("Reference array shape [time, position]:", reference.shape)
````
